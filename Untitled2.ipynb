```python
# -----------------------------
# 1. 导入库
# -----------------------------
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier
from sklearn.metrics import roc_auc_score, roc_curve
import matplotlib.pyplot as plt

# -----------------------------
# 2. 导入数据
# -----------------------------
df = pd.read_csv('train2.csv')

# 3. 分离特征和目标
X = df.drop(columns=['isDefault'])
y = df['isDefault']

# -----------------------------
# 4. 划分训练/验证/测试集 (70%/20%/10%)
# -----------------------------
X_temp, X_test, y_temp, y_test = train_test_split(
    X, y, test_size=0.1, random_state=42, stratify=y
)
val_ratio = 20 / 90  # 验证集占临时集比例
X_train, X_val, y_train, y_val = train_test_split(
    X_temp, y_temp, test_size=val_ratio, random_state=42, stratify=y_temp
)

print("训练集样本数:", X_train.shape[0])
print("验证集样本数:", X_val.shape[0])
print("测试集样本数:", X_test.shape[0])

# -----------------------------
# 5. 随机森林模型训练
# -----------------------------
rf_model = RandomForestClassifier(
    n_estimators=200,
    max_depth=None,
    random_state=42,
    n_jobs=-1
)
rf_model.fit(X_train, y_train)
y_val_rf = rf_model.predict_proba(X_val)[:,1]
auc_rf = roc_auc_score(y_val, y_val_rf)
print("随机森林验证集 AUC:", auc_rf)

# -----------------------------
# 6. XGBoost模型训练
# -----------------------------
xgb_model = XGBClassifier(
    n_estimators=200,
    learning_rate=0.1,
    max_depth=6,
    objective='binary:logistic',
    eval_metric='auc',
    use_label_encoder=False,
    n_jobs=-1,
    random_state=42
)
xgb_model.fit(X_train, y_train)
y_val_xgb = xgb_model.predict_proba(X_val)[:,1]
auc_xgb = roc_auc_score(y_val, y_val_xgb)
print("XGBoost验证集 AUC:", auc_xgb)

# -----------------------------
# 7. 绘制ROC曲线
# -----------------------------
fpr_rf, tpr_rf, _ = roc_curve(y_val, y_val_rf)
fpr_xgb, tpr_xgb, _ = roc_curve(y_val, y_val_xgb)

plt.figure(figsize=(8,6))
plt.plot(fpr_rf, tpr_rf, label=f'Random Forest AUC={auc_rf:.4f}', color='darkorange', lw=2)
plt.plot(fpr_xgb, tpr_xgb, label=f'XGBoost AUC={auc_xgb:.4f}', color='green', lw=2)
plt.plot([0,1], [0,1], linestyle='--', color='navy', lw=2)  # 随机基线
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('ROC Curve - va')
plt.legend(loc='lower right')
plt.grid(True)
plt.show()
```

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split, ParameterGrid
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier
from sklearn.metrics import roc_auc_score, roc_curve
import matplotlib.pyplot as plt

# -----------------------------
# 1. 导入数据
# -----------------------------
df = pd.read_csv('train2.csv')
X = df.drop(columns=['isDefault'])
y = df['isDefault']

# -----------------------------
# 2. 划分数据集 70/20/10
# -----------------------------
X_temp, X_test, y_temp, y_test = train_test_split(
    X, y, test_size=0.1, random_state=42, stratify=y
)
val_ratio = 20 / 90
X_train, X_val, y_train, y_val = train_test_split(
    X_temp, y_temp, test_size=val_ratio, random_state=42, stratify=y_temp
)

# -----------------------------
# 3. 随机森林调优
# -----------------------------
rf_param_grid = {
    'n_estimators': [100, 200],
    'max_depth': [None, 10, 20],
    'min_samples_split': [2, 5]
}

best_auc_rf = 0
best_params_rf = None

for params in ParameterGrid(rf_param_grid):
    rf = RandomForestClassifier(**params, random_state=42, n_jobs=-1)
    rf.fit(X_train, y_train)
    y_val_pred = rf.predict_proba(X_val)[:,1]
    auc = roc_auc_score(y_val, y_val_pred)
    if auc > best_auc_rf:
        best_auc_rf = auc
        best_params_rf = params
print("随机森林最佳参数:", best_params_rf)
print("验证集 AUC:", best_auc_rf)

# -----------------------------
# 4. XGBoost调优
# -----------------------------
xgb_param_grid = {
    'n_estimators': [100, 200],
    'max_depth': [3, 6, 10],
    'learning_rate': [0.05, 0.1]
}

best_auc_xgb = 0
best_params_xgb = None

for params in ParameterGrid(xgb_param_grid):
    xgb = XGBClassifier(
        objective='binary:logistic',
        eval_metric='auc',
        use_label_encoder=False,
        random_state=42,
        n_jobs=-1,
        **params
    )
    xgb.fit(X_train, y_train)
    y_val_pred = xgb.predict_proba(X_val)[:,1]
    auc = roc_auc_score(y_val, y_val_pred)
    if auc > best_auc_xgb:
        best_auc_xgb = auc
        best_params_xgb = params
print("XGBoost最佳参数:", best_params_xgb)
print("验证集 AUC:", best_auc_xgb)

# -----------------------------
# 5. 使用调好的参数训练最终模型
# -----------------------------
# 随机森林
rf_final = RandomForestClassifier(**best_params_rf, random_state=42, n_jobs=-1)
rf_final.fit(X_train, y_train)
y_test_rf = rf_final.predict_proba(X_test)[:,1]
auc_test_rf = roc_auc_score(y_test, y_test_rf)

# XGBoost
xgb_final = XGBClassifier(
    objective='binary:logistic',
    eval_metric='auc',
    use_label_encoder=False,
    random_state=42,
    n_jobs=-1,
    **best_params_xgb
)
xgb_final.fit(X_train, y_train)
y_test_xgb = xgb_final.predict_proba(X_test)[:,1]
auc_test_xgb = roc_auc_score(y_test, y_test_xgb)

print("最终测试集随机森林 AUC:", auc_test_rf)
print("最终测试集XGBoost AUC:", auc_test_xgb)

# -----------------------------
# 6. 绘制测试集 ROC曲线
# -----------------------------
fpr_rf, tpr_rf, _ = roc_curve(y_test, y_test_rf)
fpr_xgb, tpr_xgb, _ = roc_curve(y_test, y_test_xgb)

plt.figure(figsize=(8,6))
plt.plot(fpr_rf, tpr_rf, label=f'Random Forest AUC={auc_test_rf:.4f}', color='darkorange', lw=2)
plt.plot(fpr_xgb, tpr_xgb, label=f'XGBoost AUC={auc_test_xgb:.4f}', color='green', lw=2)
plt.plot([0,1], [0,1], linestyle='--', color='navy', lw=2)
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('ROC Curve - test')
plt.legend(loc='lower right')
plt.grid(True)
plt.show()
```
