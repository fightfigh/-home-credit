```python
import pandas as pd

# 导入 train1.csv
df = pd.read_csv('train1.csv')

# 查看前5行
df.head()
```

```python
# 查看每一列缺失值数量
df.isnull().sum()
```

```python
# 查看非数值型变量
non_numeric_cols = df.select_dtypes(exclude=['number']).columns

print(non_numeric_cols)
```

```python
import numpy as np
import pandas as pd

# 只保留：
# 1. 数值型变量
# 2. 唯一值种类 > 10

numeric_cols = [
    col for col in df.select_dtypes(include=['number']).columns
    if df[col].nunique() > 10
]

# 保存异常值结果
outlier_result = {}

# 3-sigma 异常值检测
for col in numeric_cols:
    
    mean = df[col].mean()
    std = df[col].std()
    
    lower = mean - 3 * std
    upper = mean + 3 * std
    
    # 异常值
    outliers = df[(df[col] < lower) | (df[col] > upper)]
    
    outlier_result[col] = len(outliers)

# 转成DataFrame
outlier_df = pd.DataFrame({
    '变量名': outlier_result.keys(),
    '异常值数量': outlier_result.values()
})

# 按异常值数量排序
outlier_df.sort_values(by='异常值数量', ascending=False)
```

```python
import numpy as np
import pandas as pd

# 只保留：
# 1. 数值型变量
# 2. 唯一值种类 > 10

numeric_cols = [
    col for col in df.select_dtypes(include=['number']).columns
    if df[col].nunique() > 10
]

# 复制数据集
df_clean = df.copy()

# 3-sigma 清除异常值
for col in numeric_cols:
    
    mean = df_clean[col].mean()
    std = df_clean[col].std()
    
    lower = mean - 3 * std
    upper = mean + 3 * std
    
    # 保留正常值
    df_clean = df_clean[
        (df_clean[col] >= lower) &
        (df_clean[col] <= upper)
    ]

# 查看清洗后数据集形状
print(df_clean.shape)

# 查看前5行
df_clean.head()
```

```python
from sklearn.preprocessing import LabelEncoder

# 复制数据
df_clean2 = df_clean.copy()

# 需要编码的列
cols = ['grade', 'subGrade']

# Label Encoding
le = LabelEncoder()

for col in cols:
    df_clean2[col] = le.fit_transform(df_clean2[col].astype(str))

# 删除 issueDate 列
df_clean2.drop('issueDate', axis=1, inplace=True)

# 查看结果
df_clean2.head()
```

```python
# 查看所有列名
print(df_clean2.columns.tolist())
```

```python
# 判断是否全部为数值型特征
all_numeric = all(df_clean2.dtypes.apply(lambda x: np.issubdtype(x, np.number)))

print("是否全部为数值型特征：", all_numeric)
```

```python
# 删除 id 和 policyCode 两列
df_clean2 = df_clean2.drop(['id', 'policyCode'], axis=1)

# 查看数据集
print(df_clean2.head())

# 查看数据集形状
print(df_clean2.shape)
```

```python
# 保存新的数据集
df_clean2.to_csv('train2.csv', index=False)

print("train2.csv 保存成功")
```
